```python
import google.generativeai as genai
import warnings
warnings.filterwarnings("ignore") 


API_KEY = "" 
genai.configure(api_key=API_KEY)

model_name = 'gemini-3.5-flash'
model = genai.GenerativeModel(model_name)

print("Easy promt")
prompt_1 = "Translate to ukraninan next: 'Artificial intelegence may be useful"
response_1 = model.generate_content(prompt_1)
print(f"LLMs respond:\n{response_1.text}\n")

print("Few-shot prompt")
prompt_2 = """
Determine the sentiment of the review. 
Review: "Excellent" -> Positive. 
Review: "Okay" -> Neutral. 
Review: "The code is written terribly, lots of bugs" ->
"""
response_2 = model.generate_content(prompt_2)
print(f"LLMs respond:\n{response_2.text}\n")

print("System Role (Роль викладача)")
model_with_role = genai.GenerativeModel(
    model_name,
    system_instruction="Act like a teacher in school"
)

prompt_3 = "Name three best players from Real Madrid"
response_3 = model_with_role.generate_content(prompt_3)
print(f"LLMs respond in teacher role:\n{response_3.text}\n")
```
